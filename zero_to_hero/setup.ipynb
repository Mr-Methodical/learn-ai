```python
a = torch.tensor(4.2)
b = torch.tensor([0.3, -1.2, 4.0])         # vector
c = torch.zeros(27, 27)                    # matrix of 0's
d = torch.randn(32, 3, 10)                 # 3-D tensor of random numbers
for name, t in [('a', a), ('b', b), ('c', c), ('d', d)]:
    print(f"{name}: shape {t.shape}  dims {t.dim()}  total numbers {t.numel()}")
```
